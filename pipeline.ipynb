```python
df['gender'].value_counts()


df_clean = df[df['gender'].isin(['Male','Female'])].copy()

print(f"\n Rows removed (other gender):{ len(df)- len(df_clean)}")

df_clean['gender'].value_counts()
```

```python
smoke_his = {
    'never' :'Never',
    'No Info' : 'No Info',
    'former' :'Former',
    'current' : 'Current',
    'not current' : 'Not Current',
    'ever' : 'Ever'
}

df_clean['smoking_history'] = df_clean['smoking_history'].map(smoke_his).fillna('No Info')
```

```python
len(df_clean)
df_clean = df_clean.drop_duplicates()


len(df_clean)
```
